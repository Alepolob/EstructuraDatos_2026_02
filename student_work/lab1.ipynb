```python
#A)
def p_cuadratico(a, x):
    suma = 0
    #grado = len(a)-1
    xi = 1
    for i in range(len(a)):
        suma += a[i]*xi
        xi *= x

    return suma
```

```python
#B)
def p_mejorado(a,x):
  suma = 0
  xi = 1
  for i in range(len(a)):
    suma += a[i] * xi
    xi *= x
    return suma
```

```python
#C
def horner(a, x):
    suma = a[-2] + x*a[-1]
    a1 = a[:-2]
    for i in a1[::-1]:
        suma = i+ x*suma
    return suma
```

Ejercicio 2 — División sintética

```python
def division_sintetica(a, r):
    res = a[-1]
    for i in range(len(a) - 2, -1, -1):
        res = a[i] + res * r
    return res
```

```python
a = [3, 2, 5, -1, 4]
x = 2
print("horner:              ", horner(a, x))
print("division_sintetica:  ", division_sintetica(a, x))
```

Ejercicio 3 — Comparación de algoritmos

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.linspace(1, 400, 400)
f_A = n**2 + n
f_B = (n**3) / 200

plt.figure(figsize=(8, 5))
plt.plot(n, f_A, label=r'Algoritmo A ($n^2 + n$)', color='blue')
plt.plot(n, f_B, label=r'Algoritmo B ($n^3 / 200$)', color='red')
plt.axvline(x=201, color='gray', linestyle='--', label=r'Punto de corte $n_0 \approx 201$')

plt.title('Comparación de Complejidad Temporal')
plt.xlabel('Tamaño de la entrada (n)')
plt.ylabel('Operaciones')
plt.legend()
plt.grid(True)
plt.show()
```

Ejercicio 4 — Three-way set disjointness

```python
def disjointness_nlogn(A, B, C):
    A_sort = sorted(A)
    B_sort = sorted(B)
    C_sort = sorted(C)
    i = j = k = 0
    n = len(A)
    while i < n and j < n and k < n:
        if A_sort[i] == B_sort[j] == C_sort[k]:
            return False
        min_val = min(A_sort[i], B_sort[j], C_sort[k])
        if A_sort[i] == min_val:
            i += 1
        elif B_sort[j] == min_val:
            j += 1
        else:
            k += 1
    return True
```

Ejercicio 7 — Búsqueda secuencial

```python
def busqueda_secuencial(L, x):
    for elemento in L:
        if elemento == x:
            return True
    return False
```

Ejercicio 8 — Búsqueda binaria

```python
def busqueda_binaria(L, x):
    inicio = 0
    fin = len(L) - 1
    while inicio <= fin:
        medio = (inicio + fin) // 2
        if L[medio] == x:
            return True
        elif L[medio] < x:
            inicio = medio + 1
        else:
            fin = medio - 1
    return False
```
