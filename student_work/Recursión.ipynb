1. Escribir una funcion 'potencia_dos(n)' que compute $2^n$
2. Una función recursiva potencia_dos(n) que haga lo anterior

```python
def exponencial(n):
  res=1
  for i in range(n):
    res=res*2
  return res
```

```python
exponencial(5)
```

```python
def potencia_dos_rec(n):

  if n==0:
    return 1
  else:
    a=potencia_dos_rec(n-1)
    return 2*a
```

```python
potencia_dos_rec(5)
```

```python
def potencia_dos_rec1(n):
  if n==0:
    return 1
  a=potencia_dos_rec1(n//2)
  if n%2==0:
    return a*a
  else:
    return 2*(a*a)
```

```python
potencia_dos_rec1(5)
```

```python
def maximo(L):
  # Base case: if the list has only one element, that's the maximum
  if len(L) == 1:
    return L[0]
  else:
    # Recursive step: compare the first element (L[0]) with the maximum of the rest of the list (L[1:])
    # L[1:] creates a new list (a slice), so the original list L is not modified.
    rest_of_list_max = maximo(L[1:])
    if L[0] > rest_of_list_max:
      return L[0]
    else:
      return rest_of_list_max
```

```python
print(maximo([1,2,3,41,5,9,6,7]))
```

```python
def maximo_rec(L):
  if len(L)==1:
    return L[0]
  maximoten=maximo_rec(L[::-1])
  if maximoten>L[:-1]:
    return maximoten
  else:
    return L[-1]
```

```python
maximo_rec([1,2,3,4,5,9,6,7])
```
